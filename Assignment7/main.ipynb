```python
import nltk
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
# Initial downloads for NLTK components
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# 1. Sample Document
doc = """
Far far away, behind the word mountains, far from the countries Vokalia and Consonantia,
there live the blind texts. Separated they live in Bookmarksgrove right at the coast of
the Semantics, a large language ocean. A small river named Duden flows by their place 
and supplies it with the necessary regelialia. It is a paradisematic country, in which
roasted parts of sentences fly into your mouth. Even the all-powerful Pointing has no
control about the blind texts it is an almost unorthographic life One day however a small
line of blind text by the name of Lorem Ipsum decided to
"""
```

```python
# 2. Tokenization
tokens = word_tokenize(doc.lower())
print(f"Tokens: {tokens}\n")
```

```python
# 3. POS Tagging
pos_tags = nltk.pos_tag(tokens)
print(f"POS Tags: {pos_tags[:5]}... (showing first 5)\n")
```

```python
# 4. Stop Words Removal
stop_words = set(stopwords.words('english'))
filtered_tokens = [w for w in tokens if w.isalpha() and w not in stop_words]
print(f"Filtered (No Stop Words): {filtered_tokens}\n")
```

```python
# 5. Stemming
stemmer = PorterStemmer()
stemmed = [stemmer.stem(w) for w in filtered_tokens]
print(f"Stemmed: {stemmed}\n")
```

```python
# 6. Lemmatization
lemmatizer = WordNetLemmatizer()
lemmatized = [lemmatizer.lemmatize(w) for w in filtered_tokens]
print(f"Lemmatized: {lemmatized}\n")
```

```python
# 7. TF-IDF Representation
# We treat our single sentence as a list (corpus) to calculate scores
corpus = [" ".join(lemmatized)]
vectorizer = TfidfVectorizer()
tfidf_matrix = vectorizer.fit_transform(corpus)
```

```python
# Display TF-IDF results
print("TF-IDF Representation:")
words = vectorizer.get_feature_names_out()
for i, word in enumerate(words):
    print(f"{word}: {tfidf_matrix[0, i]:.4f}")
```
