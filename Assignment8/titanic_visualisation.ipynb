```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("titanic")
```

```python
df.head()
```

```python
df.describe()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='age', bins=32)
plt.show()
```

```python
plt.figure(figsize=(8, 4))
sns.histplot(data=df, x='fare', bins=40)
plt.show()
```

```python
df['agegroup'] = pd.cut(df['age'], bins=[0, 12, 18, 35, 60, 100],
                        labels=['Child', 'Teen', 'Young Adult', 'Adult', 'Senior'])
```

```python
survivors = df[df['survived'] == 1]
```

```python
grouped = survivors.groupby(['agegroup', 'sex'], observed=True).size().reset_index(name='count')
```

```python
sns.barplot(data=grouped, x='agegroup', y='count', hue='sex')
plt.title("Number of Survivors by Age Group and Gender")
plt.show()
```
